#00 Download the dataset

Downloads MSD Task02_Heart from the mirror used by MONAI and extracts it to `data/Task02_Heart/`. The data is not stored in the repository.

source: http://medicaldecathlon.com/

```python
# you just need to run this script once to download and extract the dataset, its almost 500mb
import hashlib
import tarfile
import urllib.request
from pathlib import Path

URL = "https://msd-for-monai.s3-us-west-2.amazonaws.com/Task02_Heart.tar"
MD5 = "06ee59366e1e5124267b774dbd654057"  # fingerprint of the original archive
DATA_DIR = Path("../data")
archive = Path("../Task02_Heart.tar")

if (DATA_DIR / "Task02_Heart" / "dataset.json").exists():
    print("dataset already extracted")
else:
    if not archive.exists():
        urllib.request.urlretrieve(URL, archive)
    assert hashlib.md5(archive.read_bytes()).hexdigest() == MD5, "corrupted download"
    with tarfile.open(archive) as tar:
        tar.extractall(DATA_DIR, filter="data")
    print("extracted to", DATA_DIR / "Task02_Heart")
```
